5. Generate a synthetic dataset with correlated features. Visualize the correlation matrix of a dataset using a
heatmap.


```python
np.random.seed(42)
size = 100
feature1 = np.random.rand(size) * 10
feature2 = feature1 * 0.8 + np.random.rand(size) * 2
feature3 = feature1 * -0.5 + feature2 * 0.3 + np.random.rand(size) * 2

df = pd.DataFrame({
    'Feature 1': feature1,
    'Feature 2': feature2,
    'Feature 3': feature3
})
correlation_matrix = df.corr()
plt.figure(figsize=(6, 4))
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm', linewidths=0.5)
plt.title("Correlation Matrix Heatmap")
plt.show()
```

```python
import numpy as np
import pandas as pd
import plotly.graph_objects as go
import plotly.express as px
np.random.seed(30)
data = {
    'X': np.random.uniform(-10, 10, 300),
    'Y': np.random.uniform(-10, 10, 300),
    'Z': np.random.uniform(-10, 10, 300)
}

df = pd.DataFrame(data)
fig= go.Figure()
fig.add_trace(go.Scatter3d(x=df['X'],y=df['Y'],z=df['Z'],mode='markers',marker=dict(size=5,color='blue')))
fig.update_layout(title='3D Scatter Plot',scene=dict(xaxis_title='X',yaxis_title='Y',zaxis_title='Z'))
fig.show()
```

1.Create a Bokeh plot displaying a sine wave. Set x-values from 0 to 10 and y-values as the sine of x.

```python
import bokeh.io
import bokeh.plotting
bokeh.io.output_notebook()

from bokeh.plotting import figure, show
from bokeh.io import output_file
import numpy as np
import pandas as pd
x = np.linspace(0, 10, 100)
y = np.sin(x)

output_file("line.html")
p = figure(title="Sine Wave", x_axis_label="X axis", y_axis_label="Y axis")
p.line(x, y, legend_label="sin(x)", line_width=2, color="blue")
show(p)
```

2.Create a Bokeh scatter plot using randomly generated x and y values. Use different sizes and colors for the
markers based on the 'sizes' and 'colors' columns.

```python
from bokeh.models import ColumnDataSource
np.random.seed(42)
data = {
    'x': np.random.rand(100) * 100,
    'y': np.random.rand(100) * 100,
    'sizes': np.random.randint(5, 30, 100),
    'colors': ["#%02x%02x%02x" % (r, g, b) for r, g, b in zip(
        np.random.randint(50, 255, 100),
        np.random.randint(50, 255, 100),
        np.random.randint(50, 255, 100))]
}

df = pd.DataFrame(data)
output_file("scatter_plot.html")
source = ColumnDataSource(df)
p = figure(title="Scatter Plot", x_axis_label="X Values", y_axis_label="Y Values",width=700)
p.scatter(x='x', y='y', size='sizes', fill_color='colors', alpha=0.6, source=source)
show(p)
```

```python
data_heatmap = np.random.rand(10, 10)
x = np.linspace(0, 1, 10)
y = np.linspace(0, 1, 10)
xx, yy = np.meshgrid(x, y)

x_flat = xx.flatten()
y_flat = yy.flatten()
colors = data_heatmap.flatten()
output_file("heatmap.html")
p = figure(title="Simple Heatmap", x_axis_label="X", y_axis_label="Y")
p.rect(x_flat, y_flat, width=0.1, height=0.1, fill_color=["#%02x%02x%02x" % (int(c*255), 50, 150) for c in colors])
show(p)
```
